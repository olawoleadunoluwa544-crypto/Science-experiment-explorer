```python
import sqlite3

import pandas as pd
conn=sqlite3.connect(':memory:')
```

```python
conn.execute("DROP TABLE IF EXISTS experiment")
conn.execute("CREATE TABLE experiment(experiment_id INTEGER PRIMARY KEY, experiment_name TEXT NOT NULL, time_needed INTEGER NOT NULL)")
conn.execute("DROP TABLE IF EXISTS details")
conn.execute("CREATE TABLE details(details_id INTEGER PRIMARY KEY, experiment_id INTEGER, items_needed TEXT NOT NULL)")
```

```python
conn.executemany("INSERT OR IGNORE INTO experiment VALUES(?,?,?)", [
    (1, 'Volcano', 10),
    (2, 'Oobleck', 15),
    (3, 'Lava Lamp', 30),
    (4, 'Salt Crystals growing', 45)
])

conn.executemany("INSERT OR IGNORE INTO details VALUES(?,?,?)",[
    (1, 1, 'Baking Soda'),
    (2, 1, 'Vinegar'),
    (3, 1, 'food coloring'),
    (4, 2, 'Cornstarch'),
    (5, 2, 'Water'),
    (6, 3, 'Vegetable Oil'),
    (7, 3, 'Water'),
    (8, 3, 'Food Coloring'),
    (9, 3, 'Alka-Seltzer Tablets'),
    (10, 4, 'Table Salt'),
    (11, 4, 'Water')
])

conn.commit()
```

```python
experiment = "SELECT *  FROM experiment"
print(pd.read_sql_query(experiment, conn))
print("\n")
```

```python
query = """
SELECT experiment.experiment_id AS id FROM experiment
"""
print(pd.read_sql_query(query, conn))
```

```python
query = """
SELECT details.items_needed AS items FROM details
"""
print(pd.read_sql_query(query, conn))
```

```python
querry = """
SELECT * FROM experiment WHERE time_needed = (SELECT MIN(time_needed) FROM experiment);
"""
print(pd.read_sql_query(querry, conn))
```
